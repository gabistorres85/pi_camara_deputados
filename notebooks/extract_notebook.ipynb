```python
import sys
import os
from datetime import datetime
import pandas as pd
sys.path.append(os.path.abspath("../src"))
```

```python
from camara_deputados.extraction.api_client import CamaraAPIClient
from camara_deputados.extraction.base_extractor import BaseExtractor
from camara_deputados.extraction.write import DataWrite
```

```python
#instâncias 

api_client = CamaraAPIClient()
extractor = BaseExtractor(api_client)
writer = DataWrite()
```

```python
legislatura = '55,56,57'
df_deputado = extractor.extract(
    endpoint=f'deputados?idLegislatura={legislatura}'
)
```

```python
#salva na bronze deputados lista
writer.save_parquet(
    df=df_deputado,
    dataset='deputado_lista',
    layer='bronze'
)
```

```python
df_deputado.head()
```

```python
ids_deputados = df_deputado['id']
```

```python
# extração do detalhamento do deputado
df_deputadosDetalhamento = extractor.extract_by_ids(
    ids=ids_deputados,
    endpoint_template='deputados/{id}',
    desc='Detalhamento deputados'
)
```

```python
writer.save_parquet(
    df_deputadosDetalhamento,
    dataset='deputado_detalhamento',
    layer='bronze'
    )
```

```python
# Extração de frentes dos deputados

df_deputadoFrentes = extractor.extract_by_ids(
    ids=ids_deputados, 
    endpoint_template='deputados/{id}/frentes',
    desc='Frentes dos deputados')

```

```python
#salvar deputados e suas frentes na bronze

writer.save_parquet(
    df= df_deputadoFrentes,
    dataset = 'deputado_frentes',
    layer='bronze'
)
```

```python
# Extração de proposições
years = [2018,2019,2020,2021,2022,2023,2024,2025,2026]
df_proposicoes = extractor.extract_by_period(years, 
                                             endpoint_template='proposicoes?ano={year}',
                                             desc= 'Extração porposições por período')
```

```python
# salvando proposiçoes

writer.save_parquet(
    df=df_proposicoes,
    dataset='proposicoes',
    layer = 'bronze'
)
```
