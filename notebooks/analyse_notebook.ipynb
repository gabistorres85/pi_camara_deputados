```python
def analisar_dfs(dfs):
    resultado = []

    for nome_tabela, df in dfs.items():
        for coluna in df.columns:
            
            total = len(df)
            nulos = df[coluna].isnull().sum()
            
            resultado.append({
                "tabela": nome_tabela,
                "coluna": coluna,
                "tipo": df[coluna].dtype,
                "nulos": nulos,
                "%_nulos": nulos / total,
                "unicos": df[coluna].astype(str).nunique(),
                "total_linhas": total
            })

    return pd.DataFrame(resultado)
```

```python
# Verifica merge entre silver_autores e silver_deputados
df_autores = dfs_silver['silver_autores']
df_deputados = dfs_silver['silver_deputado']

print("Colunas em silver_autores:")
print(df_autores.columns.tolist())
print("\nColunas em silver_deputado:")
print(df_deputados.columns.tolist())

# Tenta merge por nome (normalizando)
from unidecode import unidecode

df_autores['nome_clean'] = df_autores['nom_Autor'].str.upper().str.strip().apply(lambda x: unidecode(x) if isinstance(x, str) else x)
df_deputados['nome_clean'] = df_deputados['nom_NomeCivil'].str.upper().str.strip().apply(lambda x: unidecode(x) if isinstance(x, str) else x)

# Merge
merge_result = df_autores.merge(
    df_deputados[['id_Deputado', 'nome_clean']],
    on='nome_clean',
    how='left'
)

print(f"\nTotal de autores: {len(df_autores)}")
print(f"Autores com match em deputados: {merge_result['id_Deputado'].notna().sum()}")
print(f"Autores sem match: {merge_result['id_Deputado'].isna().sum()}")

# Mostra exemplos
print("\nExemplos com match:")
print(merge_result[merge_result['id_Deputado'].notna()][['nom_Autor', 'id_Deputado']].head())
```
